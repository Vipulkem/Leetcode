```python
fruits = [3,6,1]
baskets = [6,4,7]


# fruits = [4,2,5]
# baskets = [3,5,4]

cnt = 0
for i in fruits:
    for j in range(len(baskets)):
        if i >= baskets[j]:
            baskets.pop(j)
            print("A : ",baskets)
            break
    else:
        cnt+=1
        print(baskets)
        print("cnt : ",cnt)
    if len(baskets) == 0:
        break
print(cnt)
```
