```python
class Solution:
    def zeroFilledSubarray(self, nums: List[int]) -> int:
        cnt = nums.count(0)
        for i in range(2,cnt+1):
            for j in range(len(nums)-i+1):
                if nums[j:j+i] == [0]*i:
                    cnt += 1
        return cnt
```

```python
class Solution:
    def zeroFilledSubarray(self, nums: List[int]) -> int:
        cnt = 0
        consecutive = 0
        
        for num in nums:
            if num == 0:
                consecutive += 1
                cnt += consecutive   # add all subarrays ending here
            else:
                consecutive = 0
                
        return cnt
```
