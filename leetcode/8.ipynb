```python
def t(s):
    s= "".join(s.split())
    s1=""
    for ch in s:
        if s[0] == "-":
            s1=s1+ch
            pass
        elif ch.isdigit():
            s1=s1+ch
        else:
            break
    if s1=="":
        return 0
    elif int(s1) <= (-2**31):
        return -2**31
    elif int(s1) >= 2**31-1:
        return 2**31
    else:
        return int(s1)
t("   -042")
```
