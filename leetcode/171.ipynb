```python
class Solution:
    def titleToNumber(self, columnTitle: str) -> int:
        s = 0
        for i in range(1,len(columnTitle)+1):
            if i==1:
                s += ord(columnTitle[-i])-64
            else:
                s += ((26**(i-1))*(ord(columnTitle[-i])-64))
        return s
```
