```python
class Solution:
    def maximum69Number (self, num: int) -> int:
        num = list(str(num))
        l = []
        l += [int("".join(num))]
        for i in range(len(num)):
            if num[i]=='6':
                num[i]='9'
                l += [int("".join(num))]
                num[i]='6'
                continue
                
            if num[i]=='9':
                num[i]='6'
                l += [int("".join(num))]
                num[i]='9'
                continue
        return max(l)
    
```
