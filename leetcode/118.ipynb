```python
# this is only for 1st five
def generate(numRows):
        if numRows == 1:
            return [1]
        else:
            num = 11**(numRows-1)
            l = [int(i) for i in str(num)]
            return l
```

```python
generate(5)
```

```python
def pas(numRows):
    v = [[1],[1,1]]
    
    for i in range(3,numRows+1):
        x = v[-1]
        l1 = [1]
        for j in range(i-2):
            val = x[j]+x[j+1]
            l1 += [val]
        l1 += [1]
        v.append(l1)
    print(v[:numRows])
```

```python
pas(7)
```
