```python
class Solution:
    def maxFrequencyElements(self, nums: List[int]) -> int:
        sn = list(set(nums))
        if len(nums) == len(sn):
            return len(nums)
        
        max_freq_num = sn[0]
        max_freq_count = nums.count(sn[0])

        for i in sn:
            if nums.count(i) > max_freq_count:
                max_freq_num = i
                max_freq_count = nums.count(i)


        # print("max_freq_num : ",max_freq_num)
        count = 0
        for i in sn:
            if nums.count(i) == max_freq_count:
                count  += max_freq_count
        
        return count
        # print(" max freq count : ",count)
        
```
