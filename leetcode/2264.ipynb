```python
def largestGoodInteger(num):
    l=[]
    for i in range(len(num)-2):
        if num[i]==num[i+1]==num[i+2]:
            l.append(num[i:i+3])
    if len(l)!=0:
        return max(l)
    
    return ""
    
```
