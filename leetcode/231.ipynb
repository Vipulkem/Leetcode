```python
n = 262143
for i in range(0,n//2+1):
    if 2**i==n:
        print(True)
        break
    if 2**i>n:
        break
print(False)
```

```python
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        for i in range(0,n//2+1):
            if 2**i>n:
                return False
            if 2**i==n:
                return True
            
        return False
```
