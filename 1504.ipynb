```python
mat = [[1,0,1],[1,1,0],[1,1,0]]
m, n = len(mat), len(mat[0])
heights = [0]*n
total = 0

for i in range(m):
    # update histogram heights
    for j in range(n):
        heights[j] = heights[j] + 1 if mat[i][j] == 1 else 0

    # count submatrices ending at this row using a monotonic stack
    stack = []  # each item: (height, span_count)
    row_sum = 0
    for j in range(n):
        span = 1
        while stack and stack[-1][0] >= heights[j]:
            h, c = stack.pop()
            row_sum -= h * c
            span += c
        row_sum += heights[j] * span
        stack.append((heights[j], span))
        total += row_sum

print(total)
```
