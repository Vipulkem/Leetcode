```python
def matrix_cell(cCenter,rCenter,cols,rows):
    l = []
    for col in range(cols-1,cCenter-cols,-1):
        for row in range(rows-1,rCenter-rows,-1):
            if col>=0 and row>=0:
                l +=[[row,col]]
    return l
```
