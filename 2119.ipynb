```python
num = 526
num = int(''.join(reversed(str(num))))
print(num)
num = int(''.join(reversed(str(num))))
print(num)
```

```python
class Solution:
    def isSameAfterReversals(self, num: int) -> bool:
        n1 = int(''.join(reversed(str(num))))
        n2 = int(''.join(reversed(str(n1))))
        return num==n2
        
```
