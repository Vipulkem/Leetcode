```python
def reorderedPowerOf2(n: int) -> bool:
    from collections import Counter
    
    # Digit count of n
    count_n = Counter(str(n))
    
    # Check against digit counts of all powers of 2
    for i in range(31):  # 2^0 up to 2^30 fits within 10^9
        power = 1 << i   # 2^i
        if Counter(str(power)) == count_n:
            return True
    
    return False
```

# from collections import Counter

- Stores this frequency map of digits.

- If n = 821 → count_n = Counter({'8': 1, '2': 1, '1': 1}).

- If n = 100 → count_n = Counter({'1': 1, '0': 2}).
